# Training Model using ML

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('AI_Impact_on_Jobs_2030.csv')
```

```python
df.head()
```

```python
x = df[['Years_Experience','AI_Replacement_Risk']]
y =  df['Job_Title']

model = DecisionTreeClassifier()
model.fit(x,y)

pred = model.predict([[10,0.5]])
pred
```
